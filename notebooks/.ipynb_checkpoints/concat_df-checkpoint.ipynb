```python
import pandas as pd 
import numpy as np
```

```python
network_list = [
    'conf','email_eu','hospital','school','work'
]
for network_name in network_list:
    df1 = pd.read_csv(f'../results/weighted/given_th_n=200/weighted_complex_{network_name}_EPH.csv') 
    df2 = pd.read_csv(f'../results/weighted/given_th_n=100/weighted_complex_{network_name}_EPH.csv') 
    df_merge = pd.concat([df1,df2],axis=0)
    df_merge.to_csv(f'../results/weighted/given_th_n=300/weighted_complex_{network_name}_EPH.csv')

    df1 = pd.read_csv(f'../results/weighted/given_th_n=200/weighted_simple_{network_name}_EPH.csv') 
    df2 = pd.read_csv(f'../results/weighted/given_th_n=100/weighted_simple_{network_name}_EPH.csv') 
    df_merge = pd.concat([df1,df2],axis=0)
    df_merge.to_csv(f'../results/weighted/given_th_n=300/weighted_simple_{network_name}_EPH.csv')
```

```python
network_list = [
    'conf','email_eu','hospital','school','work'
]
for network_name in network_list:
    df1 = pd.read_csv(f'../results/unweighted/given_th_n=200/unweighted_complex_{network_name}_EPH.csv') 
    df2 = pd.read_csv(f'../results/unweighted/given_th_n=100/unweighted_complex_{network_name}_EPH.csv') 
    df_merge = pd.concat([df1,df2],axis=0)
    df_merge.to_csv(f'../results/unweighted/given_th_n=300/unweighted_complex_{network_name}_EPH.csv')

    df1 = pd.read_csv(f'../results/unweighted/given_th_n=200/unweighted_simple_{network_name}_EPH.csv') 
    df2 = pd.read_csv(f'../results/unweighted/given_th_n=100/unweighted_simple_{network_name}_EPH.csv') 
    df_merge = pd.concat([df1,df2],axis=0)
    df_merge.to_csv(f'../results/unweighted/given_th_n=300/unweighted_simple_{network_name}_EPH.csv')
```
