```python
import pandas as pd 
import numpy as np 

#tqdm
from tqdm import tqdm

#visualization
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import sem

import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

import gc
```

## unweighted

```python
network_list = ['conf','email_eu','hospital','school','work']

for network_name in network_list:

    df_complex = pd.read_csv(f'../results/unweighted/given_th_n=300/unweighted_complex_{network_name}_EPH.csv')      
    df_simple = pd.read_csv(f'../results/unweighted/given_th_n=300/unweighted_simple_{network_name}_EPH.csv')
    df_complex = df_complex.sample(len(df_simple),random_state=1093)
    EPH_complex = df_complex['EPH'].tolist()
    EPH_simple = df_simple['EPH'].tolist()


    # Increase the overall figure size and DPI for better resolution
    fig, ax = plt.subplots(figsize=(8, 6), dpi=1200)
    
    # Plot histograms and KDE
    sns.histplot(EPH_complex, bins=15, kde=True, color='#377eb8', alpha=0.5, linewidth=0, ax=ax, label='Complex', stat='density')
    sns.histplot(EPH_simple, bins=15, kde=True, color='#e41a1c', alpha=0.5, linewidth=0, ax=ax, label='Simple', stat='density')
    
    # Add labels with increased font size
    ax.set_xlabel('EPH', fontsize=15)
    ax.set_ylabel('Frequency', fontsize=15)
    
    # Increase tick label size for the main plot
    ax.tick_params(axis='both', which='major', labelsize=12)
    
    # Increase legend font size
    # ax.legend(fontsize=30,loc='lower right')
    ax.legend(loc='lower right')
    
    # Create the inset axes for the box plot with increased size
    axins = ax.inset_axes([0.45, 0.45, 0.5, 0.5])
    
    # Create the box plot with simple first, then complex
    data = [EPH_simple, EPH_complex]
    bp = axins.boxplot(data, vert=1, widths=0.45)
    
    # Adding labels to the inset plot with increased font size
    axins.set_xticklabels(['Simple', 'Complex'], fontsize=15)
    axins.set_ylabel('EPH', fontsize=15)
    
    # Adjust tick label size for the inset plot
    axins.tick_params(axis='both', which='major', labelsize=12)
    
    # Tight layout for better spacing
    plt.tight_layout()
    
    # Save the figure in high resolution
    plt.savefig(f'../results/unweighted_plots/n=300_distribution_from_given_th/{network_name}_dis_unweighted.pdf', format='pdf', dpi=1200, bbox_inches='tight')
```
