```python
import networkx as nx
import os

# Parameters
N = 400     # Number of nodes
k = 14       # Each node is connected to k nearest neighbors
ps = [np.round(0.1 * i,1) for i in range(1, 10)]  # Rewiring probabilities from 0.01 to 0.09

# Output directory
output_dir = '../networks/unweighted'
os.makedirs(output_dir, exist_ok=True)

name_list = [] 
# Generate and save graphs
for p in ps:
    G = nx.watts_strogatz_graph(n=N, k=k, p=p)
    network_name = f"smallworld_p{str(p)}"
    name_list.append(network_name)
    file_path = os.path.join(output_dir, f'G_unweighted_{network_name}.graphml')
    nx.write_graphml(G, file_path)
    print(f"Saved: {file_path}")
```
